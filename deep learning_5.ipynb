```python
x = df["PetalLengthCm"].values

print("First 10 Petal Length Values:")
print(x[:10])
```

```python
x_normalized = (x - np.mean(x)) / np.std(x)

print("First 10 Normalized Values:")
print(x_normalized[:10])
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

```python
def relu(x):
    return np.maximum(0, x)
```

```python
def tanh(x):
    return np.tanh(x)
```

```python
sigmoid_output = sigmoid(x_normalized)
relu_output = relu(x_normalized)
tanh_output = tanh(x_normalized)

print("Activation functions applied successfully!")
```

```python
results = pd.DataFrame({
    "Original Petal Length": x,
    "Normalized Input": x_normalized,
    "Sigmoid Output": sigmoid_output,
    "ReLU Output": relu_output,
    "Tanh Output": tanh_output
})

print(results.head(10))
```
